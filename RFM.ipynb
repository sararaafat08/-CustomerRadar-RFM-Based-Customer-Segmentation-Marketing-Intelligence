```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=df.dropna(subset=['CustomerID'])
```

```python
df['InvoiceDate']=pd.to_datetime(df['InvoiceDate'])
```

```python
df.info()
```

```python
df=df[~df['InvoiceNo'].astype(str).str.startswith('C')]
```

```python
df = df[df['Quantity']>0]
```

```python
df['TotalPRice'] = df['Quantity'] * df['UnitPrice']
```

```python
df.head()
```

```python
#RFM analysis
```

```python
#snapshot date : the last time that the customer made a purchase ? we need 'reference date'
snapshot_date= df['InvoiceDate'].max() + pd.Timedelta(days=1)
snapshot_date
```

```python
#Build RFM table
rfm = df.groupby('CustomerID').agg({'InvoiceDate':lambda x: (snapshot_date - x.max()).days,
                                    'InvoiceNo':'nunique',
                                    'TotalPRice':'sum'
                                   })
#gm3na kul data kul customer m3 b3d , agg to make more than 1 calculation 
# (a5r 3mlit 4ra2 lel customer kant mn kam yum) InoviceDate --> Recency 
# (el customer a4tra kam mra) InvoiceNo --> Frequency 
# (el customer srf kam) TotalPrice --> Monetary 
```

```python
#rename columns
rfm.columns = ['Recency','Frequency','Monetary']
```

```python
rfm.head()
```

```python
# score ? / 10
```

```python
rfm['R_score']=pd.qcut(rfm['Recency'],5,labels=[5,4,3,2,1])
#quantile cut : by2sm el data le 5 groups mtsawya fel 3dd, kul ma recency kant a2l fa kant a7sn
# a2l Recency a3la Score
```

```python
rfm.head()
```

```python
rfm['F_score']=pd.qcut(rfm['Frequency'].rank(method='first'),5,labels=[1,2,3,4,5])
# kul ma el customer y4tri aktr kul ma kan a7sn
```

```python
rfm['M_score']=pd.qcut(rfm['Monetary'],5,labels=[1,2,3,4,5])
# kul ma el customer ysrf aktr kul ma yb2a ahm
```

```python
rfm.head()
```

```python
rfm["R_score"] = rfm["R_score"].astype(int)
rfm["F_score"] = rfm["F_score"].astype(int)
rfm["M_score"] = rfm["M_score"].astype(int)
```

```python
#score
rfm['RFM_Score']=(rfm['R_score'].astype(str)
                  + rfm['F_score'].astype(str)
                  + rfm['M_score'].astype(str)
                 )
```

```python
def segment_customer(row):
    if row['RFM_Score'] == '555':
        return 'Champions'
    elif row['R_score'] >= 4 and row['F_score'] >= 4:
        return 'Loyal Customers'
    elif row['R_score'] >= 4 and row['F_score'] <= 2:
        return 'New Customers'
    elif row['R_score'] <= 2 and row['F_score'] >= 4:
        return 'At Risk'
    elif row['R_score'] == 1 and row['F_score'] == 1:
        return 'Lost Customer'
    else:
        return 'Potential Loyalists'
rfm['Segment'] = rfm.apply(segment_customer, axis=1)
```

```python
print(rfm.columns)
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=rfm, x='Segment', order=rfm['Segment'].value_counts().index,
              hue='Segment',
              palette='ocean',
              legend=False)
plt.title('Customer Segment Distribution', fontsize=15)
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=20)
plt.show()
#which segment contain more customers?
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=rfm,
    x="Segment",
    y="Monetary",
    hue="Segment",
    palette="crest",
    legend=False
)

plt.title("Average Spending by Customer Segment")

plt.xlabel("Customer Segment")

plt.ylabel("Average Monetary Value")

plt.xticks(rotation=20)

plt.show()
#who spends more?
```

```python
plt.figure(figsize=(8,5))
plt.hist(rfm["Monetary"],
    bins=30,
    color="steelblue",
    edgecolor="black"
)
plt.title("Distribution of Customer Spending")
plt.xlabel("Monetary")
plt.ylabel("Frequency")
plt.show()
#more customers spends ?
```

```python
plt.figure(figsize=(7,6))

sns.heatmap(
    rfm[["Recency","Frequency","Monetary"]].corr(),
    annot=True,
    cmap="ocean",
    linewidths=0.5
)
plt.title("Correlation Matrix")
plt.show()
#there's a relation between montary and freq and recency?
```
